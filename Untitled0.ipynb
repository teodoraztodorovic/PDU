```python
def load_image_and_mask(case_id):
    # slika
    dcm_path = os.path.join(dicom_dir, dicom_map[case_id])
    dcm = pydicom.dcmread(dcm_path)
    img = dcm.pixel_array.astype(np.float32)
    img = (img - img.min()) / (img.max() - img.min())  # normalizacija u [0,1]

    # maska
    xml_path = os.path.join(xml_dir, case_id + '.xml')
    with open(xml_path, 'rb') as f:
        plist = plistlib.load(f)
    rois = plist['Images'][0]['ROIs']

    mask = np.zeros(dcm.pixel_array.shape, dtype=np.uint8)
    for roi in rois:
        if roi.get('Name') == 'Mass':
            pts = np.array([eval(p) for p in roi['Point_px']], dtype=np.int32)
            cv2.fillPoly(mask, [pts], 1)

    return img, mask
```

```python
import torch
from torch.utils.data import Dataset
import cv2

class INBreastMassDataset(Dataset):
    def __init__(self, ids, dicom_map, xml_dir, dicom_dir, img_size=512, transform=None):
        self.ids = ids
        self.dicom_map = dicom_map
        self.xml_dir = xml_dir
        self.dicom_dir = dicom_dir
        self.img_size = img_size
        self.transform = transform

    def __len__(self):
        return len(self.ids)

    def __getitem__(self, idx):
        case_id = self.ids[idx]
        img, mask = load_image_and_mask(case_id)  # funkcija koju smo već napravili

        img = cv2.resize(img, (self.img_size, self.img_size), interpolation=cv2.INTER_LINEAR)
        mask = cv2.resize(mask, (self.img_size, self.img_size), interpolation=cv2.INTER_NEAREST)

        img_t = torch.from_numpy(img).unsqueeze(0).float()      # (1, H, W)
        mask_t = torch.from_numpy(mask).unsqueeze(0).float()    # (1, H, W)

        if self.transform:
            img_t, mask_t = self.transform(img_t, mask_t)

        return img_t, mask_t
```

```python
import segmentation_models_pytorch as smp

model = smp.UnetPlusPlus(
    encoder_name='resnet34',
    encoder_weights='imagenet',
    in_channels=1,
    classes=1,
    activation=None
)
```

```python
loss_fn = smp.losses.TverskyLoss(mode='binary', alpha=0.3, beta=0.7)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)

# brza provera forward pass-a
imgs, masks = next(iter(train_loader))
imgs, masks = imgs.to(device), masks.to(device)

preds = model(imgs)
loss = loss_fn(preds, masks)
print('Output shape:', preds.shape, '| Loss:', loss.item())
```

```python
import torch.optim as optim

optimizer = optim.Adam(model.parameters(), lr=1e-4)

def compute_metrics(preds, masks, threshold=0.5):
    preds_bin = (torch.sigmoid(preds) > threshold).float()
    tp = (preds_bin * masks).sum()
    fp = (preds_bin * (1 - masks)).sum()
    fn = ((1 - preds_bin) * masks).sum()

    recall = tp / (tp + fn + 1e-7)
    dice = (2 * tp) / (2 * tp + fp + fn + 1e-7)
    return recall.item(), dice.item()

def run_epoch(loader, model, loss_fn, optimizer=None):
    is_train = optimizer is not None
    model.train() if is_train else model.eval()

    total_loss, total_recall, total_dice = 0, 0, 0
    n_batches = len(loader)

    for imgs, masks in loader:
        imgs, masks = imgs.to(device), masks.to(device)

        with torch.set_grad_enabled(is_train):
            preds = model(imgs)
            loss = loss_fn(preds, masks)

            if is_train:
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

        recall, dice = compute_metrics(preds, masks)
        total_loss += loss.item()
        total_recall += recall
        total_dice += dice

    return total_loss / n_batches, total_recall / n_batches, total_dice / n_batches


num_epochs = 20
for epoch in range(num_epochs):
    train_loss, train_recall, train_dice = run_epoch(train_loader, model, loss_fn, optimizer)
    val_loss, val_recall, val_dice = run_epoch(val_loader, model, loss_fn)

    print(f'Epoch {epoch+1}/{num_epochs} | '
          f'Train loss: {train_loss:.4f} recall: {train_recall:.4f} dice: {train_dice:.4f} | '
          f'Val loss: {val_loss:.4f} recall: {val_recall:.4f} dice: {val_dice:.4f}')
```

```python
def evaluate_at_threshold(loader, model, threshold):
    model.eval()
    total_recall, total_dice, n = 0, 0, 0
    with torch.no_grad():
        for imgs, masks in loader:
            imgs, masks = imgs.to(device), masks.to(device)
            preds = torch.sigmoid(model(imgs))
            preds_bin = (preds > threshold).float()

            tp = (preds_bin * masks).sum()
            fp = (preds_bin * (1 - masks)).sum()
            fn = ((1 - preds_bin) * masks).sum()

            recall = (tp / (tp + fn + 1e-7)).item()
            dice = ((2*tp) / (2*tp + fp + fn + 1e-7)).item()
            total_recall += recall
            total_dice += dice
            n += 1
    return total_recall/n, total_dice/n

model.load_state_dict(torch.load('best_model.pth'))

for t in [0.2, 0.3, 0.4, 0.5]:
    r, d = evaluate_at_threshold(val_loader, model, t)
    print(f'Threshold {t}: recall={r:.4f}, dice={d:.4f}')
```

```python
# 1. Keširanje
cache = {}
for cid in all_ids:
    img, mask = load_image_and_mask(cid)
    img = cv2.resize(img, (512, 512), interpolation=cv2.INTER_LINEAR)
    mask = cv2.resize(mask, (512, 512), interpolation=cv2.INTER_NEAREST)
    cache[cid] = (img.astype(np.float32), mask.astype(np.uint8))

print('Keširano:', len(cache), 'slika')

# Nova Dataset klasa koja koristi keš
class INBreastMassDatasetCached(Dataset):
    def __init__(self, ids, cache, transform=None):
        self.ids = ids
        self.cache = cache
        self.transform = transform

    def __len__(self):
        return len(self.ids)

    def __getitem__(self, idx):
        case_id = self.ids[idx]
        img, mask = self.cache[case_id]

        if self.transform:
            augmented = self.transform(image=img, mask=mask)
            img, mask = augmented['image'], augmented['mask']

        img_t = torch.from_numpy(img).unsqueeze(0).float()
        mask_t = torch.from_numpy(mask).unsqueeze(0).float()
        return img_t, mask_t
```

```python
# 2. i 3. Izmenjena run_epoch funkcija sa AMP-om
scaler = torch.cuda.amp.GradScaler()

def run_epoch_amp(loader, model, loss_fn, optimizer=None):
    is_train = optimizer is not None
    model.train() if is_train else model.eval()
    total_loss, total_recall, total_dice, n_batches = 0, 0, 0, len(loader)

    for imgs, masks in loader:
        imgs, masks = imgs.to(device), masks.to(device)

        with torch.set_grad_enabled(is_train):
            with torch.cuda.amp.autocast():
                preds = model(imgs)
                loss = loss_fn(preds, masks)

            if is_train:
                optimizer.zero_grad()
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()

        recall, dice = compute_metrics(preds, masks)
        total_loss += loss.item()
        total_recall += recall
        total_dice += dice

    return total_loss / n_batches, total_recall / n_batches, total_dice / n_batches
```

```python
%%time
test_id = all_ids[0]
train_ids_loo = [x for x in all_ids if x != test_id]

train_ds_loo = INBreastMassDatasetCached(train_ids_loo, cache, transform=train_transform)
test_ds_loo = INBreastMassDatasetCached([test_id], cache)

train_loader_loo = DataLoader(train_ds_loo, batch_size=4, shuffle=True)
test_loader_loo = DataLoader(test_ds_loo, batch_size=1, shuffle=False)

model_loo = smp.UnetPlusPlus(encoder_name='resnet34', encoder_weights='imagenet', in_channels=1, classes=1, activation=None).to(device)
optimizer_loo = optim.Adam(model_loo.parameters(), lr=1e-4)
scheduler_loo = optim.lr_scheduler.ReduceLROnPlateau(optimizer_loo, mode='min', factor=0.5, patience=5)
loss_fn_loo = smp.losses.TverskyLoss(mode='binary', alpha=0.2, beta=0.8)

for epoch in range(25):
    train_loss, _, _ = run_epoch_amp(train_loader_loo, model_loo, loss_fn_loo, optimizer_loo)
    scheduler_loo.step(train_loss)

test_loss, test_recall, test_dice = run_epoch_amp(test_loader_loo, model_loo, loss_fn_loo)
print(f'recall: {test_recall:.4f} | dice: {test_dice:.4f}')
```

```python
from sklearn.model_selection import KFold

kf = KFold(n_splits=20, shuffle=True, random_state=42)
all_ids_arr = np.array(all_ids)

results_path = '/content/drive/MyDrive/PDU/kfold_results20.csv'

if os.path.exists(results_path):
    kfold_df = pd.read_csv(results_path)
    done_folds = set(kfold_df['fold'])
else:
    kfold_df = pd.DataFrame(columns=['fold', 'id', 'recall', 'dice'])
    done_folds = set()

fixed_epochs = 30 # malo više nego kod LOOCV pošto sad imamo manje foldova ukupno

for fold_idx, (train_idx, test_idx) in enumerate(kf.split(all_ids_arr)):
    if fold_idx in done_folds:
        continue

    train_ids_fold = all_ids_arr[train_idx].tolist()
    test_ids_fold = all_ids_arr[test_idx].tolist()

    train_ds_fold = INBreastMassDatasetCached(train_ids_fold, cache, transform=train_transform)
    test_ds_fold = INBreastMassDatasetCached(test_ids_fold, cache)

    train_loader_fold = DataLoader(train_ds_fold, batch_size=4, shuffle=True)
    test_loader_fold = DataLoader(test_ds_fold, batch_size=4, shuffle=False)

    model_fold = smp.UnetPlusPlus(encoder_name='resnet34', encoder_weights='imagenet', in_channels=1, classes=1, activation=None).to(device)
    optimizer_fold = optim.Adam(model_fold.parameters(), lr=1e-4)
    scheduler_fold = optim.lr_scheduler.ReduceLROnPlateau(optimizer_fold, mode='min', factor=0.5, patience=5)
    loss_fn_fold = smp.losses.TverskyLoss(mode='binary', alpha=0.2, beta=0.8)

    for epoch in range(fixed_epochs):
        train_loss, _, _ = run_epoch_amp(train_loader_fold, model_fold, loss_fn_fold, optimizer_fold)
        scheduler_fold.step(train_loss)

    test_loss, test_recall, test_dice = run_epoch_amp(test_loader_fold, model_fold, loss_fn_fold)

    new_row = pd.DataFrame([{'fold': fold_idx, 'id': str(test_ids_fold), 'recall': test_recall, 'dice': test_dice}])
    kfold_df = pd.concat([kfold_df, new_row], ignore_index=True)
    kfold_df.to_csv(results_path, index=False)

    print(f'Fold {fold_idx+1}/20 | recall: {test_recall:.4f} | dice: {test_dice:.4f}')

print('\nGotovo. Prosečan recall:', kfold_df['recall'].mean(), '(std:', kfold_df['recall'].std(), ')')
print('Prosečan dice:', kfold_df['dice'].mean(), '(std:', kfold_df['dice'].std(), ')')
```
